## Part 0 — Python Basics (the language itself)

Before touching data, let's learn the building blocks of Python.

### 0.1 — Printing things
The `print()` command shows something on the screen.

```python
print("Hello world!")
print("My name is Umar and I am learning Python")
```

### 0.2 — Variables
A variable is just a **box with a name** where we store a value.

```python
name = "Ali"
age = 21
height = 5.9

print(name)
print(age)
print(height)
```

### 0.3 — Doing math
Python is a powerful calculator.

```python
a = 10
b = 3

print("Add:", a + b)
print("Subtract:", a - b)
print("Multiply:", a * b)
print("Divide:", a / b)
print("Power (a to the b):", a ** b)
```

### 0.4 — Data types
Every value has a *type*. Use `type()` to check it.

```python
print(type(5))         # int  -> whole number
print(type(5.5))       # float -> decimal number
print(type("hello"))   # str  -> text
print(type(True))      # bool -> True / False
```

### 0.5 — Lists
A **list** holds many values together, inside square brackets `[ ]`.

```python
fruits = ["apple", "banana", "mango", "orange"]

print(fruits)
print("First fruit:", fruits[0])    # counting starts at 0 !
print("Second fruit:", fruits[1])
print("Last fruit:", fruits[-1])
print("How many fruits:", len(fruits))
```

```python
# We can add and change items
fruits.append("grapes")   # add to the end
print(fruits)

fruits[0] = "strawberry"  # change the first item
print(fruits)
```

### 0.6 — Dictionaries
A **dictionary** stores data as `key : value` pairs, inside curly braces `{ }`.

```python
student = {
    "name": "Sara",
    "age": 20,
    "city": "Lahore"
}

print(student)
print("Name is:", student["name"])
print("Age is:", student["age"])
```

### 0.7 — If / Else (making decisions)
The computer chooses what to do based on a condition.

```python
marks = 75

if marks >= 50:
    print("Pass 🎉")
else:
    print("Fail")
```

```python
# Try changing 'marks' above to 30 and run again!
temperature = 35

if temperature > 30:
    print("It is hot, drink water!")
elif temperature > 15:
    print("Nice weather")
else:
    print("It is cold, wear a jacket")
```

### 0.8 — Loops (do something many times)
A `for` loop repeats an action for each item.

```python
for fruit in fruits:
    print("I like", fruit)
```

```python
# Loop over numbers using range()
for i in range(1, 6):     # 1, 2, 3, 4, 5
    print("Number:", i, " | Square:", i * i)
```

### 0.9 — Functions (reusable blocks)
A function is a mini-program we can use again and again.

```python
def greet(person):
    return "Assalam-o-Alaikum, " + person + "!"

print(greet("Ahmed"))
print(greet("Fatima"))
```

```python
def area_of_rectangle(length, width):
    return length * width

print("Area:", area_of_rectangle(5, 3))
print("Area:", area_of_rectangle(10, 2))
```

## Part 3 — Selecting and Filtering Data

### 3.1 — Pick one column

```python
df["species"]
```

### 3.2 — Pick a few columns

```python
df[["species", "island", "body_mass_g"]].head()
```

### 3.3 — Pick rows by position with `.iloc`

```python
# rows 0 to 4 (first five)
df.iloc[0:5]
```

### 3.4 — Filtering: only the heavy penguins
We keep rows where a condition is `True`.

```python
heavy = df[df["body_mass_g"] > 5000]
print("Number of penguins heavier than 5000g:", heavy.shape[0])
heavy.head()
```

### 3.5 — Filtering with two conditions
Use `&` for AND. Each condition needs `( )` around it.

```python
adelie_biscoe = df[(df["species"] == "Adelie") & (df["island"] == "Biscoe")]
adelie_biscoe.head()
```

---
## Part 4 — Handling Missing Values

Real data is messy! Some cells are empty (shown as `NaN`). Let's find and fix them.

### 4.1 — Count missing values in each column

```python
df.isnull().sum()
```

### 4.2 — Drop rows that have missing values
This removes any row with at least one empty cell. We save the clean version as `df_clean`.

```python
df_clean = df.dropna()
print("Before cleaning:", df.shape[0], "rows")
print("After cleaning:", df_clean.shape[0], "rows")
```

### 4.3 — Confirm no missing values remain

```python
df_clean.isnull().sum()
```

---
## Part 5 — Exploring and Summarizing Data

From here on we use the clean table `df_clean`.

### 5.1 — Count how many of each species
`.value_counts()` is super useful.

```python
df_clean["species"].value_counts()
```

### 5.2 — How many penguins on each island?

```python
df_clean["island"].value_counts()
```

### 5.3 — Average body mass of the whole dataset

```python
avg_mass = df_clean["body_mass_g"].mean()
print("Average body mass:", round(avg_mass, 1), "grams")
```

### 5.4 — Group by species and get the average of each measurement
`groupby` is one of the most powerful tools in pandas.

```python
df_clean.groupby("species").mean(numeric_only=True)
```

### 5.5 — Sort penguins from heaviest to lightest

```python
df_clean.sort_values("body_mass_g", ascending=False).head()
```

### 5.6 — Create a NEW column
Let's convert body mass from grams to kilograms.

```python
df_clean["body_mass_kg"] = df_clean["body_mass_g"] / 1000
df_clean[["species", "body_mass_g", "body_mass_kg"]].head()
```

## Part 6 — Visualizing the Data 📊

Charts make patterns easy to see. This is the fun part!

### 6.1 — Bar chart: how many of each species
Great for counting categories.

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df_clean, x="species")
plt.title("Number of Penguins by Species")
plt.show()
```

### 6.2 — Histogram: distribution of flipper length
Shows how a single number is spread out.

```python
plt.figure(figsize=(7, 4))
plt.hist(df_clean["flipper_length_mm"], bins=20, color="skyblue", edgecolor="black")
plt.title("Distribution of Flipper Length")
plt.xlabel("Flipper length (mm)")
plt.ylabel("Count")
plt.show()
```

### 6.3 — Scatter plot: bill length vs bill depth
Each dot is one penguin. Colors show the species — watch the groups appear!

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df_clean, x="bill_length_mm", y="bill_depth_mm", hue="species")
plt.title("Bill Length vs Bill Depth")
plt.show()
```

### 6.4 — Box plot: body mass for each species
Shows the typical value and the spread, side by side.

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=df_clean, x="species", y="body_mass_g")
plt.title("Body Mass by Species")
plt.show()
```

### 6.5 — Bar chart of average body mass per species
We combine `groupby` (from Part 5) with a chart.

```python
avg_by_species = df_clean.groupby("species")["body_mass_g"].mean()

plt.figure(figsize=(7, 4))
avg_by_species.plot(kind="bar", color=["#FF8C42", "#3498DB", "#2ECC71"])
plt.title("Average Body Mass by Species")
plt.ylabel("Body mass (g)")
plt.xticks(rotation=0)
plt.show()
```

### 6.6 — Correlation heatmap
Shows which measurements move together. Closer to 1 = strong link.

```python
plt.figure(figsize=(7, 5))
corr = df_clean.select_dtypes("number").corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Between Measurements")
plt.show()
```

### 6.7 — Pairplot: everything at once! ✨
This one chart compares every pair of measurements, colored by species. Always a crowd-pleaser.

```python
sns.pairplot(df_clean, hue="species")
plt.show()
```
